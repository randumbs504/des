```python

DES:
import java.util.Scanner;
import java.util.Base64;
import javax.crypto.Cipher;
import javax.crypto.spec.SecretKeySpec;
public class DES{
public static void main(String[] args)throws Exception{
Scanner sc = new Scanner(System.in);
System.out.print("Enter a message: ");
String msg = sc.nextLine();
String Key = "12345678";
SecretKeySpec ks = new SecretKeySpec(Key.getBytes(),"DES");
Cipher c = Cipher.getInstance("DES");
c.init(Cipher.ENCRYPT_MODE,ks);
byte[] enc = c.doFinal(msg.getBytes());
String encd = Base64.getEncoder().encodeToString(enc);
System.out.print("\nEncryption: "+ encd);
c.init(Cipher.DECRYPT_MODE,ks);
byte[] dec = c.doFinal(Base64.getDecoder().decode(encd));
System.out.print("\nDecryption: "+ new String(dec));
}
}
Enter a message: HELLO
Encryption: i2oUQS3liQM=
Decryption: HELLO
```

```python
CAESAR:
#include<iostream>
#include<cctype>
using namespace std;
string caesar(string s,int k){
string r = "";
for(char ch : s){
if(isalpha(ch)){
char base = isupper(ch) ? 'A' : 'a';
r+=((ch - base + k + 26)%26 + base);
}
else{
r+=ch;
}
}
return r;
}
int main(){
string text;
int k;
cout<<"Enter the text: ";
cin>>text;
cout<<"\nEnter the key: ";
cin>>k;
string enc = caesar(text,k);
string dec = caesar(enc,-k);
cout<<"\nEncrypted: "<<enc;
cout<<"\nDecrypted: "<<dec;
}
```

```python
ELAGAMAL:
#include<iostream>
using namespace std;
long long power(long long a,long long b,long long mod){
long long r=1;
while(b){
if(b%2)
r=(r*a)%mod;
a=(a*a)%mod;
b/=2;
}
return r;
}
int main(){
long long p,g,x,msg,k;
cout<<"Enter p: ";
cin>>p;
cout<<"Enter g: ";
cin>>g;
cout<<"Enter private key: ";
cin>>x;
long long h=power(g,x,p);
cout<<"Public Key: ("<<p<<","<<g<<","<<h<<")";
cout<<"\nEnter message: ";
cin>>msg;
cout<<"Enter random key k: ";
cin>>k;
long long c1=power(g,k,p);
long long c2=(power(h,k,p)*msg)%p;
cout<<"\nEncrypted: ("<<c1<<","<<c2<<")";
long long s=power(c1,x,p);
long long dec=(c2*power(s,p-2,p))%p;
cout<<"\nDecrypted: "<<dec;
}
Enter p: 43
Enter g: 6
Enter private key: 13
Public Key: (43,6,7)
Enter message: 10
Enter random key k: 9
Encrypted: (1,10)
Decrypted: 10
```

```python
PLAYFAIR:
#include <iostream>
#include <cctype>
using namespace std;
char M[5][5] = {
{'M','O','N','A','R'},
{'C','H','Y','B','D'},
{'E','F','G','I','K'},
{'L','P','Q','S','T'},
{'U','V','W','X','Z'}
};
void pos(char ch, int &r, int &c)
{
if (ch == 'J')
ch = 'I';
for (int i = 0; i < 5; i++)
{
for (int j = 0; j < 5; j++)
{
if (M[i][j] == ch)
{
r = i;
c = j;
return;
}
}
}
}
string format(string s)
{
string t = "";
for (char c : s)
{
if (isalpha(c))
t += toupper(c);
}
for (int i = 0; i < t.length(); i += 2)
{
if (i + 1 == t.length() || t[i] == t[i + 1])
t.insert(i + 1, "X");
}
if (t.length() % 2)
t += 'X';
return t;
}
string encrypt(string s)
{
s = format(s);
string out = "";
for (int i = 0; i < s.length(); i += 2)
{
int r1, c1, r2, c2;
pos(s[i], r1, c1);
pos(s[i + 1], r2, c2);
if (r1 == r2)
{
out += M[r1][(c1 + 1) % 5];
out += M[r2][(c2 + 1) % 5];
}
else if (c1 == c2)
{
out += M[(r1 + 1) % 5][c1];
out += M[(r2 + 1) % 5][c2];
}
else
{
out += M[r1][c2];
out += M[r2][c1];
}
}
return out;
}
string decrypt(string s)
{
string out = "";
for (int i = 0; i < s.length(); i += 2)
{
int r1, c1, r2, c2;
pos(s[i], r1, c1);
pos(s[i + 1], r2, c2);
if (r1 == r2)
{
out += M[r1][(c1 + 4) % 5];
out += M[r2][(c2 + 4) % 5];
}
else if (c1 == c2)
{
out += M[(r1 + 4) % 5][c1];
out += M[(r2 + 4) % 5][c2];
}
else
{
out += M[r1][c2];
out += M[r2][c1];
}
}
return out;
}
int main()
{
string text;
cout << "Enter text: ";
cin >> text;
string enc = encrypt(text);
string dec = decrypt(enc);
cout << "Encrypted text: " << enc << endl;
cout << "Decrypted text: " << dec << endl;
return 0;
}
Enter text: INSTRUMENT
Encrypted text: GATLMZCLRQ
Decrypted text: INSTRUMENT
```

```python
HILL:
#include <iostream>
using namespace std;
int mod26(int x)
{
x %= 26;
if (x < 0)
x += 26;
return x;
}
void multiply(int key[2][2], int p[2], int c[2])
{
for (int i = 0; i < 2; i++)
{
c[i] = 0;
for (int j = 0; j < 2; j++)
{
c[i] += key[i][j] * p[j];
}
c[i] = mod26(c[i]);
}
}
void inverseKey(int key[2][2], int inv[2][2])
{
int det = mod26(key[0][0] * key[1][1] - key[0][1] * key[1][0]);
int detInv;
for (int i = 1; i < 26; i++)
{
if (mod26(det * i) == 1)
detInv = i;
}
inv[0][0] = mod26(key[1][1] * detInv);
inv[0][1] = mod26(-key[0][1] * detInv);
inv[1][0] = mod26(-key[1][0] * detInv);
inv[1][1] = mod26(key[0][0] * detInv);
}
string process(string text, int key[2][2])
{
if (text.length() % 2 != 0)
text += 'X';
string result = "";
for (int i = 0; i < text.length(); i += 2)
{
int p[2] = {text[i] - 'A', text[i + 1] - 'A'};
int c[2];
multiply(key, p, c);
result += char(c[0] + 'A');
result += char(c[1] + 'A');
}
return result;
}
int main()
{
int key[2][2] = {{3,3},{2,5}};
int inv[2][2];
string text;
cout << "Enter UPPERCASE text: ";
cin >> text;
string cipher = process(text, key);
cout << "Encrypted: " << cipher << endl;
inverseKey(key, inv);
string plain = process(cipher, inv);
cout << "Decrypted: " << plain << endl;
return 0;
}
Enter UPPERCASE text: HELLO
Encrypted: HIOZHN
Decrypted: HELLOX
```
